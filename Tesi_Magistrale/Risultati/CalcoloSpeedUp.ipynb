```python
import matplotlib.pyplot as plt

# Usa i font LaTeX se disponibili (coerenza con la tesi); se non hai una
# distribuzione LaTeX in locale, commenta il blocco rcParams sottostante.
plt.rcParams.update({
    "text.usetex": False,   # metti True se hai LaTeX installato in locale
    "font.family": "serif",
    "font.size": 11,
})

domains = ["500x500", "1000x1000", "2000x2000"]
x = range(len(domains))

speedup_cupy = [0.9, 0.7, 1.8]
speedup_raw  = [1.6, 1.1, 2.7]

fig, ax = plt.subplots(figsize=(6, 4))

ax.plot(x, speedup_cupy, marker='o', color='#1f77b4', label='FIM CuPy / FIM CPU')
ax.plot(x, speedup_raw,  marker='s', color='#d62728', label='FIM RawKernel / FIM CPU')

# Linea di riferimento a speedup = 1 (nessun guadagno rispetto alla CPU)
ax.axhline(y=1, color='gray', linestyle='--', linewidth=1)

ax.set_xticks(list(x))
ax.set_xticklabels(domains)
ax.set_xlabel("Dimensione del dominio")
ax.set_ylabel("Speedup (in secondi)")
ax.set_title("Speedup di FIM su GPU rispetto a FIM CPU")
ax.legend()
ax.grid(True, linestyle=':', linewidth=0.5)

fig.tight_layout()
```
